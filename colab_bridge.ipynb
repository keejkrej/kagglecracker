# 🚀 Antigravity Colab GPU Bridge (via tmate)

Run the cell below. It sets up an instant SSH terminal tunnel (via tmate) with no Cloudflare or WebSocket friction.

```python
# @title Start Instant GPU SSH Tunnel
import os, subprocess, time

print("[*] Installing and launching tmate tunnel...")
os.system("apt-get update -qq && apt-get install -y tmate -qq")
os.system("rm -f /tmp/tmate.sock")
os.system("tmate -S /tmp/tmate.sock new-session -d")
os.system("tmate -S /tmp/tmate.sock wait tmate-ready")

ssh_cmd = subprocess.check_output("tmate -S /tmp/tmate.sock display -p '#{tmate_ssh}'", shell=True).decode().strip()
print("\n" + "=" * 60)
print("🎉 GPU TUNNEL READY!")
print(f"{ssh_cmd}")
print("=" * 60)
print("\n👉 Copy the line above (e.g. ssh abcxyz@lon1.tmate.io) and paste it here!")

# Keep alive loop
while True:
    time.sleep(60)
```
